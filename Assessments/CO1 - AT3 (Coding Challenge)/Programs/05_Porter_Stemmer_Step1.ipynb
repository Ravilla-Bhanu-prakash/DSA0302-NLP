```python

input_data = """8
caresses
ponies
cats
feed
agreed
motoring
hopping
happy"""

data = input_data.splitlines()
Q = int(data[0])

def is_consonant(word, i):
    ch = word[i]
    if ch in "aeiou":
        return False
    if ch == 'y':
        if i == 0:
            return True
        return not is_consonant(word, i - 1)
    return True

def measure(word):
    m = 0
    i = 0
    while i < len(word) and is_consonant(word, i):
        i += 1
    while i < len(word):
        while i < len(word) and not is_consonant(word, i):
            i += 1
        if i < len(word):
            m += 1
            while i < len(word) and is_consonant(word, i):
                i += 1
    return m

def has_vowel(word):
    return any(not is_consonant(word, i) for i in range(len(word)))

def double_consonant(word):
    return (
        len(word) >= 2
        and word[-1] == word[-2]
        and is_consonant(word, len(word) - 1)
    )

def cvc(word):
    if len(word) < 3:
        return False
    return (
        is_consonant(word, len(word) - 3)
        and not is_consonant(word, len(word) - 2)
        and is_consonant(word, len(word) - 1)
        and word[-1] not in "wxy"
    )

def stem(word):
    # Step 1a
    if word.endswith("sses"):
        word = word[:-2]
    elif word.endswith("ies"):
        word = word[:-2]
    elif word.endswith("ss"):
        pass
    elif word.endswith("s"):
        word = word[:-1]

    # Step 1b
    removed = False

    if word.endswith("eed"):
        base = word[:-3]
        if measure(base) > 0:
            word = base + "ee"
    elif word.endswith("ed") and has_vowel(word[:-2]):
        word = word[:-2]
        removed = True
    elif word.endswith("ing") and has_vowel(word[:-3]):
        word = word[:-3]
        removed = True

    if removed:
        if word.endswith(("at", "bl", "iz")):
            word += "e"
        elif double_consonant(word) and word[-1] not in "lsz":
            word = word[:-1]
        elif measure(word) == 1 and cvc(word):
            word += "e"

    # Step 1c
    if word.endswith("y") and has_vowel(word[:-1]):
        word = word[:-1] + "i"

    return word

for word in data[1:Q + 1]:
    print(stem(word))
```
