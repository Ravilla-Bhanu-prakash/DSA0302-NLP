**Task 1 – Frame the Problem: Models, Algorithms and Levels of Analysis**

```python
import re

records = [
    "Patient PT-10452 admitted on 12-03-2025. BP 120/80 mmHg.",
    "Patient P-20567 has fever and was given Paracetamol 500 mg BD.",
    "Patient PT-30456 has fracture and temperature 38 C."
]

def process_record(text):
    print("\nOriginal Record:", text)

    patient_id = re.findall(r"\b(?:PT|P)[- ]?\d{4,8}\b", text)

    tokens = re.findall(r"\b\w+(?:-\w+)*\b", text)

    normalised_tokens = [word.lower() for word in tokens]

    return patient_id, normalised_tokens

for record in records:
    ids, tokens = process_record(record)

    print("Patient ID:", ids)
    print("Processed Tokens:", tokens)
```

**Task 2 – Basic Regular Expression Patterns**

```python
import re

text = """
Patient PT-10452 was admitted on 12-03-2025.
Discharged on 15/03/2025.
Prescribed Paracetamol 500 mg.
Take BD.
Hb 13.5 g/dL.
BP 120/80 mmHg.
Temperature 38 C.
ICD code A12.3.
Ward 2 Bed 14.
Contact +91 9876543210.
Age 58 years.
Symptoms for 3 days.
"""

patterns = {
    "Patient ID": r"\b(?:PT|P)[- ]?\d{4,8}\b",

    "Admission/Discharge Date": r"\b(?:0?[1-9]|[12]\d|3[01])[-/](?:0?[1-9]|1[0-2])[-/](?:19|20)\d{2}\b",

    "Drug and Strength": r"\b[A-Za-z][A-Za-z-]*\s+\d+(?:\.\d+)?\s?(?:mg|mcg|g|ml|IU)\b",

    "Dosage Frequency": r"\b(?:OD|BD|TDS|QID|once daily|twice daily|every \d+ hours)\b",

    "Laboratory Value": r"\b(?:Hb|WBC|RBC|glucose|creatinine)\s*[:=]?\s*\d+(?:\.\d+)?\s?(?:mg/dL|g/dL|mmol/L)?\b",

    "Blood Pressure": r"\b\d{2,3}\s*/\s*\d{2,3}\s?mmHg\b",

    "Temperature": r"\b\d{2,3}(?:\.\d+)?\s?°?\s?(?:C|F)\b",

    "ICD Code": r"\b[A-TV-Z]\d{2}(?:\.\d{1,4})?\b",

    "Ward and Bed": r"\b(?:Ward|W)\s*[- ]?\d{1,3}\s*[,/-]?\s*(?:Bed|B)\s*[- ]?\d{1,4}\b",

    "Contact Number": r"\b(?:\+?\d{1,3}[- ]?)?(?:\d[- ]?){10}\b",

    "Age": r"\b(?:age\s*)?\d{1,3}\s?(?:years?|yrs?|y/o)\b",

    "Symptom Duration": r"\b\d+\s?(?:hours?|days?|weeks?|months?|years?)\b"
}

for entity, pattern in patterns.items():
    result = re.findall(pattern, text, re.IGNORECASE)
    print(entity, ":", result)
```

**Task 3 – Regular Expressions for Tokenisation and Normalisation**

```python
import re

text = """
FRACTURE patient has BP 120/80 mmHg.
Paracetamol 500mg BD was prescribed.
Patient has follow-up after 3.5 days.
HTN is also reported.
"""

# Tokenisation pattern
pattern = r"(?:[A-Za-z]+(?:-[A-Za-z]+)+)|(?:[A-Za-z]+(?:\.[A-Za-z]+)+\.?)|(?:\d+(?:\.\d+)?(?:mg|mcg|g|ml|mmol/L|mmHg)?)|(?:[A-Za-z]+)|(?:[^\w\s])"

tokens = re.findall(pattern, text)

# Normalisation dictionary
abbreviations = {
    "bp": "blood pressure",
    "bd": "twice daily",
    "htn": "hypertension"
}

normalised_tokens = []

for token in tokens:
    word = token.lower()

    if word in abbreviations:
        word = abbreviations[word]

    normalised_tokens.append(word)

print("Original Text:")
print(text)

print("\nTokens:")
print(tokens)

print("\nNormalised Tokens:")
print(normalised_tokens)
```

**Task 4 – Finite State Automata (DFA, NFA and Subset Construction)**

```python
# DFA for patient ID
def validate_patient_id(value):
    state = 0

    for ch in value:
        if state == 0 and ch in "PT":
            state = 1
        elif state == 1 and ch == "-":
            state = 2
        elif state == 2 and ch.isdigit():
            state = 3
        elif state == 3 and ch.isdigit():
            state = 3
        else:
            return False

    return state == 3 and value[-1].isdigit() and 4 <= len(value.split("-")[-1]) <= 8


# NFA-style validator for contact number
def validate_contact(value):
    digits = ""

    for ch in value:
        if ch.isdigit():
            digits += ch
        elif ch not in "+ -":
            return False

    return len(digits) == 10 or (len(digits) == 12 and digits.startswith("91"))


print("DFA Validation:")
print("PT-10452:", validate_patient_id("PT-10452"))
print("P-20567:", validate_patient_id("P-20567"))
print("PT-12:", validate_patient_id("PT-12"))

print("\nNFA Validation:")
print("9876543210:", validate_contact("9876543210"))
print("+91 9876543210:", validate_contact("+91 9876543210"))
print("12345:", validate_contact("12345"))


# Subset construction example for NFA: a|b
nfa = {
    "q0": {"a": {"q1"}, "b": {"q2"}},
    "q1": {},
    "q2": {}
}

dfa_states = {
    "q0": {"q0"},
    "q1": {"q1"},
    "q2": {"q2"}
}

print("\nSubset Construction:")
for state, subset in dfa_states.items():
    print(state, "=", subset)
```

**Task 5 – Inflectional Morphology**

```python
# Irregular word forms
irregular = {
    "children": "child",
    "men": "man",
    "women": "woman",
    "teeth": "tooth",
    "feet": "foot",
    "was": "be",
    "were": "be",
    "written": "write"
}

def analyse_word(word):
    word = word.lower()

    if word in irregular:
        return irregular[word], "Irregular form"

    if word.endswith("ies"):
        return word[:-3] + "y", "Plural"

    if word.endswith("ing"):
        root = word[:-3]
        if root.endswith(root[-1:]*2) and len(root) > 2:
            root = root[:-1]
        elif root.endswith("at") or root.endswith("it"):
            pass
        return root, "Progressive"

    if word.endswith("ed"):
        root = word[:-2]
        if root.endswith(root[-1:]*2) and len(root) > 2:
            root = root[:-1]
        return root, "Past tense"

    if word.endswith("es"):
        return word[:-2], "Plural / third person"

    if word.endswith("s"):
        return word[:-1], "Plural / third person"

    return word, "Base form"


words = [
    "fractures", "fractured", "fracturing",
    "patients", "treats", "treated",
    "admitting", "children", "written"
]

for word in words:
    root, feature = analyse_word(word)
    print(word, "->", root, ":", feature)
```

**Task 6 – Derivational Morphology**

```python
# Derivational prefixes and suffixes
prefixes = {
    "hyper": "excessive or high",
    "hypo": "low or deficient",
    "nephro": "kidney",
    "cardio": "heart",
    "re": "again",
    "un": "not"
}

suffixes = {
    "itis": "inflammation",
    "ectomy": "surgical removal",
    "ology": "field of study",
    "osis": "medical condition",
    "aemia": "blood condition",
    "ation": "noun formation",
    "ment": "noun formation",
    "able": "capable of",
    "ive": "adjective formation",
    "al": "related to",
    "ic": "related to",
    "ist": "person associated with",
    "ian": "person or relation",
    "ous": "adjective formation"
}

def analyse_derivation(word):
    word = word.lower()
    found = False

    for prefix, meaning in prefixes.items():
        if word.startswith(prefix):
            print("Word:", word)
            print("Prefix:", prefix)
            print("Meaning:", meaning)
            found = True
            break

    for suffix, meaning in suffixes.items():
        if word.endswith(suffix):
            print("Word:", word)
            print("Suffix:", suffix)
            print("Meaning:", meaning)
            found = True
            break

    if not found:
        print("Word:", word)
        print("No listed derivational affix found.")

words = [
    "arthritis",
    "nephrectomy",
    "cardiology",
    "fibrosis",
    "treatment",
    "treatable",
    "preventive",
    "renal",
    "cardiac",
    "hypertension",
    "hypoglycaemia",
    "reassess",
    "untreated"
]

for word in words:
    analyse_derivation(word)
    print()
```

**Task 7 – Finite-State Morphological Parsing**

```python
lexicon = {
    "fractures": ("fracture", "N", "PL"),
    "patients": ("patient", "N", "PL"),
    "treated": ("treat", "V", "PAST"),
    "treating": ("treat", "V", "PROG"),
    "children": ("child", "N", "PL"),
    "was": ("be", "V", "PAST")
}

def analyse(word):
    word = word.lower()
    if word in lexicon:
        lemma, category, feature = lexicon[word]
        return lemma + " +" + category + " +" + feature
    return "No analysis found"

def generate(lemma, category, feature):
    for word, analysis in lexicon.items():
        if analysis == (lemma, category, feature):
            return word
    return "No word found"

words = ["fractures", "patients", "treated", "treating", "children", "was"]

for word in words:
    print(word, ":", analyse(word))

print("Generated word:", generate("fracture", "N", "PL"))
```

**Task 8 – Orthographic Rules and Two-Level Morphology**

```python
def morphology(word, suffix):
    if suffix == "ing":
        if word.endswith("e"):
            word = word[:-1]
        return word + suffix

    if suffix == "ed":
        if word.endswith("e"):
            return word + "d"
        if word.endswith("mit"):
            return word + word[-1] + suffix
        return word + suffix

    if suffix == "es":
        if word.endswith("y") and len(word) > 1 and word[-2] not in "aeiou":
            return word[:-1] + "ies"
        if word.endswith(("s", "x", "z", "ch", "sh")):
            return word + "es"
        return word + "s"

    return word + suffix

tests = [
    ("operate", "ing"),
    ("admit", "ed"),
    ("injury", "es"),
    ("box", "es"),
    ("treat", "ed")
]

for word, suffix in tests:
    print(word, "+", suffix, "=", morphology(word, suffix))
```

**Task 11 – Compare, Recommend and Defend the Normalisation Design**

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

# Sample hospital records
records = [
    "Patient has fractures",
    "Patient has fracture",
    "Patient was treated",
    "Patient is treating",
    "Patient has infection"
]

# Search query
query = "fracture"

# Method 1: No Normalisation
print("1. No Normalisation")
for record in records:
    if query in record.lower().split():
        print(record)

# Method 2: Porter Stemming
print("\n2. Porter Stemming")
query_stem = ps.stem(query)

for record in records:
    words = record.lower().split()
    stems = [ps.stem(word) for word in words]

    if query_stem in stems:
        print(record)

# Method 3: Finite-State Morphology
print("\n3. Finite-State Morphology")

lexicon = {
    "fracture": ("fracture", "N", "SG"),
    "fractures": ("fracture", "N", "PL"),
    "treated": ("treat", "V", "PAST"),
    "treating": ("treat", "V", "PROG")
}

for record in records:
    words = record.lower().split()

    for word in words:
        if word in lexicon:
            lemma = lexicon[word][0]
            if lemma == query:
                print(record)
                break
```

**Task 12 – OCR and Romanised Regional Language**

```python
import re

text = """PT-1O452
dis-
charge
patient ku fever irundhuchu"""

# Join broken words
text = re.sub(r"-\s*\n\s*", "", text)

# Correct OCR error in patient ID
text = re.sub(r"(?<=PT-)1O(?=\d{3})", "10", text)

# Romanised Tamil word list
tamil_words = ["ku", "irundhuchu", "kuduthen"]

print("Corrected Text:")
print(text)

print("\nLanguage Identification:")
for word in re.findall(r"[A-Za-z]+", text.lower()):
    if word in tamil_words:
        print(word, "- Romanised Tamil")
    else:
        print(word, "- English / Other")
```
