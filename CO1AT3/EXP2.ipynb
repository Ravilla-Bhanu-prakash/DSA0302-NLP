```python

import re

input_data = """5 3
b a !
0
1 4
5
0 b 1
1 a 2
2 a 3
3 a 3
3 ! 4
5
baa!
baaa!
ba!
b
baa"""

data = input_data.splitlines()
i = 0

n, k = map(int, data[i].split())
i += 1
alphabet = data[i].split()
i += 1
start = int(data[i])
i += 1
accept_data = list(map(int, data[i].split()))
i += 1
accepting = set(accept_data[1:])
t = int(data[i])
i += 1

transitions = {}
for _ in range(t):
    p, c, r = data[i].split()
    i += 1
    transitions.setdefault((int(p), c), set()).add(int(r))

q = int(data[i])
i += 1

for _ in range(q):
    string = data[i]
    i += 1
    if string == "#":
        string = ""
    current = {start}
    for symbol in string:
        next_states = set()
        for state in current:
            next_states.update(transitions.get((state, symbol), set()))
        current = next_states
    print("ACCEPT" if current & accepting else "REJECT")
```
