```python

input_data = """3
goose N+PL geese
go V+PAST went
good ADJ+COMP better
8
cat N+PL
church N+PL
baby N+PL
stop V+PROG
make V+PROG
carry V+PAST
goose N+PL
go V+PAST"""

data = input_data.splitlines()
i = 0

L = int(data[i])
i += 1

lexicon = {}

for _ in range(L):
    stem, feature, form = data[i].split()
    lexicon[(stem, feature)] = form
    i += 1

Q = int(data[i])
i += 1

vowels = set("aeiou")

def doublable(stem):
    if len(stem) < 3:
        return False
    if stem[-1] in "wxy":
        return False
    if not (stem[-3] not in vowels and stem[-2] in vowels and stem[-1] not in vowels):
        return False

    groups = sum(
        1 for j in range(len(stem))
        if stem[j] in vowels and (j == 0 or stem[j - 1] not in vowels)
    )
    return groups == 1

def inflect(stem, feature):
    if (stem, feature) in lexicon:
        return lexicon[(stem, feature)]

    if feature in ("N+PL", "V+3SG"):
        if stem.endswith(("s", "x", "z", "ch", "sh")):
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("o") and stem[-2] not in vowels:
            return stem + "es"
        if len(stem) >= 2 and stem.endswith("y") and stem[-2] not in vowels:
            return stem[:-1] + "ies"
        return stem + "s"

    if feature == "V+PROG":
        if len(stem) >= 3 and stem.endswith("e") and len(stem) >= 2 and stem[-2] not in vowels:
            return stem[:-1] + "ing"
        if doublable(stem):
            return stem + stem[-1] + "ing"
        return stem + "ing"

    suffixes = {
        "V+PAST": "ed",
        "ADJ+COMP": "er",
        "ADJ+SUP": "est"
    }

    suffix = suffixes[feature]

    if stem.endswith("e"):
        return stem + suffix[1:]
    if len(stem) >= 2 and stem.endswith("y") and stem[-2] not in vowels:
        return stem[:-1] + "i" + suffix
    if doublable(stem):
        return stem + stem[-1] + suffix
    return stem + suffix

for _ in range(Q):
    stem, feature = data[i].split()
    i += 1
    print(inflect(stem, feature))
```
